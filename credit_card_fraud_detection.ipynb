```python
df2 = df.fillna(df.mean())
```

```python
df2.isna().sum()
```

```python
from sklearn.model_selection import train_test_split
```

```python
y = df2['Class'].astype(int)
x = df2.drop(['Class', 'Time'], axis=1)
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.20, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression

# Create the model instance
model = LinearRegression()

# Train the model on the training data
model.fit(x_train, y_train)
```

```python
# Predict the target values for the test features
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import mean_absolute_error, r2_score

# Calculate metrics
print("R2 Score (Model Accuracy):", r2_score(y_test, y_pred))
print("Mean Absolute Error (Average Error): $", mean_absolute_error(y_test, y_pred))
```

```python
sample_row = x_test.iloc[[0]]

predicted_output = model.predict(sample_row)

print(f"Predicted Output: {predicted_output[0]:.2f}")
print(f"Actual Expected Output: {y_test.iloc[0]:.2f}")
```

```python
# 1. Find the index numbers where the transaction is actually fraud
fraud_indices = y_test[y_test == 1].index

if len(fraud_indices) > 0:

    fraud_sample = x_test.loc[[fraud_indices[0]]]

    fraud_pred = model.predict(fraud_sample)

    print(f"Predicted Output for Fraud: {fraud_pred[0]:.2f}")
    print(f"Actual Expected Output: {y_test.loc[fraud_indices[0]] :.2f}")
else:
    print("No fraud cases found in this specific test split partition!")
```

```python
from sklearn.linear_model import LogisticRegression

clf_model = LogisticRegression(max_iter=1000)
clf_model.fit(x_train, y_train)

fraud_sample = x_test.loc[[fraud_indices[0]]]

class_pred = clf_model.predict(fraud_sample)

prob_pred = clf_model.predict_proba(fraud_sample)[:, 1]

print(f"Predicted Class Label: {class_pred[0]}")
print(f"Probability of being Fraud: {prob_pred[0]*100:.2f}%")
print(f"Actual Expected Output: {y_test.loc[fraud_indices[0]]}")
```

```python
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix

all_predictions = clf_model.predict(x_test)


y_test_clean = np.round(y_test).astype(int)
all_predictions_clean = np.round(all_predictions).astype(int)


print("--- Confusion Matrix ---")
print(confusion_matrix(y_test_clean, all_predictions_clean))

print("\n--- Classification Report ---")
print(classification_report(y_test_clean, all_predictions_clean))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Generate a visual heatmap
cm = confusion_matrix(y_test_clean, all_predictions_clean)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Normal', 'Fraud'], yticklabels=['Normal', 'Fraud'])
plt.ylabel('Actual')
plt.xlabel('Predicted')
plt.title('Fraud Detection Confusion Matrix')
plt.show()
```

```python
import joblib

joblib.dump(clf_model, 'fraud_detector_model.pkl')
print("Model saved successfully as 'fraud_detector_model.pkl'!")
```

```python
import gradio as gr
import pandas as pd
import joblib

def detect_fraud_from_file(file):
    if file is None:
        return "Please upload a valid CSV transaction file.", None

    try:
        # Load your trained model
        model = joblib.load('fraud_detector_model.pkl')

        # Read the uploaded CSV data
        df = pd.read_csv(file.name)

        # 1. FIX: Get the exact 29 feature names your model was trained on
        try:
            expected_feature_names = list(model.feature_names_in_)
            # Filter the uploaded file to only include those 29 columns
            features_df = df[expected_feature_names]
        except AttributeError:
            # Fallback: If feature names weren't saved, look for 'Time' and 'V1' to 'V28'
            cols = ['Time'] + [f'V{i}' for i in range(1, 29)]
            if all(c in df.columns for c in cols):
                features_df = df[cols]
            else:
                # Absolute fallback: just slice the first 29 columns if names don't match
                features_df = df.iloc[:, :29]

        # Hard check to ensure dimensions match perfectly now
        if features_df.shape[1] != 29:
            return f"❌ Mismatch Error! The model expects 29 features, but input features has {features_df.shape[1]} columns.", None

        # Run predictions on a small preview sample to save memory/speed
        preview_df = df.head(100).copy()
        preview_features = features_df.head(100)

        predictions = model.predict(preview_features)
        probabilities = model.predict_proba(preview_features)[:, 1]

        # Format the preview table outputs
        preview_df.insert(0, 'Fraud Probability (%)', (probabilities * 100).round(2))
        preview_df.insert(0, 'Risk Assessment', ['🚨 FRAUD' if p == 1 else '✅ NORMAL' for p in predictions])

        # Run a quick check on the full file for totals
        full_preds = model.predict(features_df)
        fraud_count = (full_preds == 1).sum()

        summary_text = f"📊 Scan Completed!\nTotal Scanned: {len(df)} rows.\nFlagged Critical Items: {fraud_count} fraud cases detected.\n(Displaying preview of first 100 rows below)"

        return summary_text, preview_df

    except Exception as e:
        return f"❌ System Error: {str(e)}", None

# --- Re-Launch Layout ---
with gr.Blocks(theme=gr.themes.Soft()) as demo:
    gr.Markdown("# 💳 AI Credit Card Fraud Detection Interface")
    file_input = gr.File(label="Upload Transactions CSV", file_types=[".csv"])
    submit_btn = gr.Button("🚀 Run AI Scan", variant="primary")
    status_output = gr.Textbox(label="System Status & Analytics Summary")
    data_output = gr.Dataframe(label="Flagged Analytics Report Preview")

    submit_btn.click(fn=detect_fraud_from_file, inputs=file_input, outputs=[status_output, data_output])

demo.launch(share=True, prevent_thread_lock=True)
```
